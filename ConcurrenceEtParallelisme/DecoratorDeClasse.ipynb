```python

def log(self):
    print(self.nom)

def log_decorator(cls):
    cls.log = log
    return cls

@log_decorator
class MyClass:
    def __init__(self, nom):
        self.nom = nom


myclass = MyClass("MArtin")
```

```python
myclass.log()
```

```python
for a in myclass:
    print(a)
```

```python
with myclass:
    print(myclass)
```
