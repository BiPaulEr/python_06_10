```python
import time, threading
def printer(caractere, temps_att):
    print(threading.current_thread())
    for i in range(0, 10):
        print(caractere, flush = True, end='')
        time.sleep(temps_att)
    return ("Je suis ", caractere, temps_att)
print(threading.current_thread())
t1 = Thread(target=printer, args=("*", 1), name="JAck")
print(t1)
t1.start()
print(t1)
t2 = Thread(target=printer,  args=("$", 2))
t2.start()
t1.join()
t2.join()
print("end")
print(t1)
print(t2)
```

```python
import threading
from threading import Thread, current_thread
from random import randint

def show_value(nom, data):
    try:
        val = data.value
    except AttributeError:
        print(nom, current_thread().name, ' - No value yet')
    else:
        print(nom, current_thread().name, ' - value =', val)
        
def worker(data_local, data_shared):    
    show_value("LOCAL", data_local)
    show_value("SHARED", data_shared)
    data_local.value = randint(1, 100)
    data_shared.value = randint(1, 100)
    show_value("LOCAL", data_local)
    show_value("SHARED", data_shared)
    
print(current_thread().name, ' - Starting')

local = threading.local()
local.value = -99
show_value("LOCAL", local)

class SharedData:
    pass
shared = SharedData()
shared.value = -99
show_value("SHARED", shared)


for i in range(2):
    t = Thread(name='W' + str(i), target=worker, args=(local, shared))
    t.start()
    
show_value("LOCAL", local)
print(current_thread().name, ' - Done')
```
