```python
prenoms = ["Paul", "Ernest", "Martin"]
noms = ["Dupont", "Aigouyi", "Lefbvre", "Taille"]
gen = zip(prenoms, noms)

prenoms.append("QUIOIE")
for prenom, nom in gen:
    print( prenom, nom )
```

```python
for index, prenom in enumerate(prenoms):
    print(index, prenom)
print(list(enumerate(prenoms)))
```

```python
liste = []
print(list(enumerate(zip(prenoms, noms))))
for index, (prenom, nom) in enumerate(zip(prenoms, noms)):
    liste.append( index * (prenom + nom))
print(liste)
    
```
