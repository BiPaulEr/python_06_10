```python

class Pyramide:
    def __init__(self, longeur_base, hauteur):
        self.longeur_base = longeur_base
        self.hauteur = hauteur
    @property
    def aire_base(self):
        return self.longeur_base**2
    @property
    def volume(self):
        return (1/3) * self.aire_base * self.hauteur
    @property
    def aire_base(self):
        return self.longeur_base**2
    @property
    def perimeter(self):
        return 4*self.longeur_base + 4* math.sqrt(((math.sqrt(2) * self.aire_base) / 2) + self.hauteur**2)

    @property
    def longeur_base(self):
        return self.__longeur_base
    @longeur_base.setter
    def longeur_base(self, valeur):
        if valeur < 0:
            raise ValueError
        self.__longeur_base = valeur

    @property
    def hauteur(self):
        return self.__hauteur
    @hauteur.setter
    def hauteur(self, valeur):
        if valeur < 0:
            raise ValueError
        self.__hauteur = valeur
    
        


pyramide = Pyramide(4, 6)
print(f"Périmètre de la base : {pyramide.perimeter}") 
print(f"Aire de la base : {pyramide.aire_base}")
print(f"Volume de la pyramide : {pyramide.volume}") 
```
