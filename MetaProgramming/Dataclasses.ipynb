```python
class Vector:
     def __init__(self, x, y, z):
         self.x = x
         self.y = y
         self.z = z
     def __repr__(self):
         return f"<Vector: x={self.x}, y={self.y}, z={self.z}>"
     def __eq__(self, other):
         return self.x == other.x and self.y == other.y and self.z == other.z

     def __gt__(self, other):
        return self.x > other.x
     def __ge__(self, other):
        return self.x >= other.x
        
```

```python
vector1= Vector(3, 6, 8)
vector2 = Vector(7, 6, 8)
vector3 = Vector(3, 6, 8)
```

```python
print(vector1)
```

```python
vector1 == vector3
```

```python
vector1 > vector3
```

```python
vector1 >= vector3
```

```python
from dataclasses import dataclass, field

@dataclass(order = True, frozen = True)
class Vector3D:
    prenom: str = field( compare = False)
    x : int = field( repr = False)
    y : int = field( repr = False)
    z : int = field( repr = False)

vector1= Vector3D("Pol",2, 6, 8)
vector2 = Vector3D("Al",7, 6, 8)
vector3 = Vector3D("OK",3, 6, 8)
print(vector1)
print(vector1 == vector3)
```

```python
vector1 >= vector3
```

```python
vector1 < vector3
```

```python
from dataclasses import dataclass, field
v = 5
@dataclass
class Produit:
    nom: str 
    prix : int = v
def func(produit):
    return produit.prix
@dataclass
class Courses:
    liste : list[Produit] = field(default_factory=list)
    prix_courses : int = field(init=False)
    def __post_init__(self):
        self.prix_courses = sum(map(lambda produit: produit.prix, self.liste))
        self.prix_courses = sum((produit.prix for produit in self.liste))

courses = Courses([Produit("Soupe", 5), Produit("Poireaux", 6), Produit("banana", 7)])
print(courses)
```
