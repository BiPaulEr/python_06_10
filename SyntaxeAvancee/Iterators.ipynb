**Reprise pourquoi __iter__ ieratorable -> iterator**

```python
class Count:
    def __init__(self, compteur):
        self.compteur = compteur 
        self.init = compteur
    def __iter__(self):
        self.compteur = self.init
        return self
    def __next__(self):
        if self.compteur <= 0:
            raise StopIteration
        self.compteur = self.compteur -1
        return self.compteur 

c = Count(3)
```

```python
for a in c:
    print(a)
```

```python
for a in c:
    print(a)
```

```python
class CountIterable:
    def __init__(self, compteur):
        self.compteur = compteur 
    def __next__(self):
        if self.compteur <= 0:
            raise StopIteration
        self.compteur = self.compteur -1
        return self.compteur 
        
class CountIterator:
    def __init__(self, init):
        self.init = init
    def __iter__(self):
        return CountIterable(self.init)

c = CountIterator(3)
for a in c:
    print(a)
for a in c:
    print(a)
```
