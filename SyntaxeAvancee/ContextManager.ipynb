```python
class ContextManager():
    def __enter__(self):
        print("ENTREE DS LE CONTEXTE")
        return "OBJECT"
    def __exit__(self, type_exc, instance_exc, traceback):
        if isinstance(type_exc, OSError):
             print("osEROOOR ", instance_exc)
        if isinstance(type_exc, ValueError): 
             print("VALLLLLUE ", instance_exc)
        if type_exc is None:
            print("ON A LA FIN DE  LHISTOIRE")
        print("SORTIE DU CONTEXTE")
        
with ContextManager() as quelquechose:
    print(quelquechose)
    #raise OSError("OUI")
    print("dans le contexte")
```

**Vrai exemple**

```python
import requests
try:
    session = requests.Session()
    headers = {'Accept': 'text/plain'}
    response = session.get("http://icanhazdadjoke.com", headers=headers)
    print(response.text)
finally:
    session.close()
```

```python
class SessionManager():
    def __enter__(self):
        self.session = requests.Session()
        return self.session
    def __exit__(self, t_e, i_e, tb):
        self.session.close()
        
with SessionManager() as sm:
    headers = {'Accept': 'text/plain'}
    response = sm.get("http://icanhazdadjoke.com", headers=headers)
    print(response.text)
    
```

**supress**

```python
from contextlib import suppress

with suppress(FileNotFoundError), open("fichier12.txt") as file:
    for line in file.readlines():
        print(line)
```
