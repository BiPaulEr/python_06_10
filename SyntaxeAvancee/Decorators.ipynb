```python
import time
def benchmark(function):
    def wrapper(*args, **kwargs):
        begin = time.time()
        result = function(*args, **kwargs)
        end = time.time() - begin
        print(function.__name__, "benchmark -> ", end)
        return result
    return wrapper

@benchmark
def function():
    time.sleep(2)
    return 5

print(function())
```

```python
@benchmark
def function_5():
    time.sleep(5)
    return 5
print(function_5())
```

**Decorateur avec arguments**

```python
def call_decorator_arg(nombre_call = 3):   
    def decorator(function):
        def wrapper():
            for i in range(0, nombre_call):
                function()
        return wrapper
    return decorator
    

@call_decorator_arg(3)
def function1():
    print("Je suis focntion1")
```

```python
function1()
```

```python
@call_decorator_arg(5)
def function2():
    print("Je suis focntion2")
function2()
```

```python
def types_correcteur(liste_types):
    def type_correcteur(function):
        def wrapper(*args):
            good = True
            if len(liste_types) != len(args):
                good = False
                print("PAS MEME NOMBRE")
            for index, types in enumerate(liste_types):
                if not isinstance(args[index], liste_types[index]):
                    good = False
                    print(" NON  ", type(args[index]), "attendu", liste_types[index])
            if good:
                return function(*args)
        return wrapper
    return type_correcteur
    
@types_correcteur([int])
def majeur(a):
    return a >= 18

majeur("A")  
```

```python
import functools

def simple_decorator(function):
    """ decorateur dociumentation"""
    @functools.wraps(function)
    def wrapper():
        print("Avant execution")
        function()
        print("Apres execution")
    return wrapper

@simple_decorator
def function():
    """la documentation de function qui dit oui"""
    return "OUI"
@simple_decorator
def function2():
    """la documentation de function qui dit non"""
    return "NON"
    
print(function.__doc__)
print(function2.__doc__)
print(simple_decorator.__doc__)
```
